Titanic Passengers Data Extraction and Cleaning

## Import Libraries

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

## Data Exteraction

```python
import csv
Dataset = pd.read_csv('titanic.csv')
```

## Dataset Show

```python
Dataset
```

## Passengers Survival and Death

```python
survival_count = Dataset['Survived'].value_counts()
ax = survival_count.plot(kind='bar', edgecolor='black')
ax.bar_label(ax.containers[0])
plt.xticks([0,1], ["Died", "Survived"], rotation=0)
plt.title("Passenger Survival Rate")
plt.ylabel("Number of passenger onboard")
plt.tight_layout()
```

## Passenger Class Distribution 

```python
Class_Distribution = Dataset['Pclass'].value_counts().sort_index()
ax = Class_Distribution.plot(kind='bar', edgecolor='black', figsize=(6,4))
ax.bar_label(ax.containers[0])
plt.xlabel("Passenger Class")
plt.title("Passenger Class Distribution")
plt.ylabel("Number of passenger onboard")
plt.tight_layout()
```

## Passenger Age

```python
Age_Distribution = Dataset['Age'].fillna(0)
plt.figure(figsize=(6,4))
ax = Age_Distribution.plot(kind='hist', bins=40, edgecolor='black')
ax.bar_label(ax.containers[0], fontsize=8)
plt.xlabel("Age")
plt.ylabel("Number of Passengers onboard")
plt.title("Passenger Age Distribution")
plt.tight_layout()
```

## Passenger Gender 

```python
Gender_count = Dataset['Sex'].value_counts()
plt.figure(figsize=(6,4))
plt.pie(Gender_count, labels=Gender_count.index, startangle=90, wedgeprops={'edgecolor':'black'}, autopct='%1.1f%%')
plt.title('Passenger Gender')
plt.tight_layout()
```

## Number Of Sibiling or Spouses Aboard Titanic

```python
SibSp_Count = Dataset['Siblings/Spouses Aboard'].value_counts().sort_index()
plt.figure(figsize=(6,4))
ax = SibSp_Count.plot(kind='bar', edgecolor='black')
ax.bar_label(ax.containers[0])
plt.xlabel("Number of Siblings/Spouses aboard")
plt.title("Siblings/Spouses Aboard Distribution")
plt.ylabel("Number of passenger onboard")
plt.tight_layout()
```

## Number Of Parents Or Children Aboard

```python
ParCh_Count = Dataset['Parents/Children Aboard'].value_counts().sort_index()
plt.figure(figsize=(6,4))
ax = ParCh_Count.plot(kind='bar', edgecolor='black')
ax.bar_label(ax.containers[0])
plt.xlabel("Number of Parents/Children aboard")
plt.title("Parents/Children Aboard Distribution")
plt.ylabel("Number of passenger onboard")
plt.tight_layout()
```

## Passenger Fare

```python
Fare_Count = Dataset['Fare'].fillna(0)
plt.figure(figsize=(6,4))
ax = Fare_Count.plot(kind='hist', bins=40, edgecolor='black')
ax.bar_label(ax.containers[0], fontsize=8)
plt.xlabel("Fare Price ($)")
plt.ylabel("Number of Passengers")
plt.title("Passenger Fare Distribution")
plt.tight_layout()
```
